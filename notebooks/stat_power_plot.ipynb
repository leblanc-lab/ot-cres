```python
import numpy as np
import matplotlib.pyplot as plt

```

```python
# Define epsilon range
eps = np.linspace(0.0, 0.4, 500)

# Define scaling relation
ratio = 1 / (1 - 2 * eps)**2

# Create figure
fig, ax = plt.subplots(figsize=(8, 6))

# Main curve
ax.plot(
    eps,
    ratio,
    color="tomato",
    linewidth=2,
    label=r"$\frac{N(\epsilon)}{N(0)} = \frac{1}{(1 - 2\epsilon)^2}$"
)

# Horizontal reference line at 1
ax.axhline(
    1,
    color="black",
    linestyle="--",
    linewidth=2
)

# Vertical reference line at epsilon = 0.375
frac = 0.375-(0.25*0.375)
ax.axvline(
    frac,
    color="mediumturquoise",
    linestyle="--",
    linewidth=2
)

# Labels
ax.set_xlabel(r"Negative Fraction $\epsilon$", fontsize=24)
ax.set_ylabel(r"$N(\epsilon)/N(0)$", fontsize=24)

# Axis limits
ax.set_xlim(-0.02, 0.42)
ax.set_ylim(0, 26)

# Tick formatting
ax.tick_params(axis='both', labelsize=14)

# Legend
ax.legend(fontsize=20, loc="upper left", frameon=True)

# Tight layout
plt.tight_layout()

# Save figure
plt.savefig("negative_fraction_scaling.png", dpi=300)

# Show plot
plt.show()
```
