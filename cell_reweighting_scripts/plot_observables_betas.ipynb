```python
import numpy as np
import ROOT
import matplotlib.pyplot as plt
import pyhepmc
from array import array
```

```python
#identify hard process events from 100k sample that are all 0s
points = np.load('/oscar/data/mleblan6/rjain/ppzjj_100k/hardprocess_points.npy')
mask = np.where(np.all(points == 0, axis=(1, 2)))[0]

print(mask)
```

```python
#100K SAMPLE ARRAYS

big_weight = np.load('/oscar/data/mleblan6/rjain/ppzjj_100k/weight_100k.npy') * 1.455 * 10**4
big_ht = np.load('/oscar/data/mleblan6/rjain/ppzjj_100k/ppzjj_NLO_100k_ht.npy')

big_zpt = np.load('/oscar/data/mleblan6/rjain/ppzjj_100k/ppzjj_NLO_100k_zcoords.npy')[:,0]
big_zeta = np.load('/oscar/data/mleblan6/rjain/ppzjj_100k/ppzjj_NLO_100k_zcoords.npy')[:,1]
big_zphi = np.load('/oscar/data/mleblan6/rjain/ppzjj_100k/ppzjj_NLO_100k_zcoords.npy')[:,2]

big_ljetpt = np.load('/oscar/data/mleblan6/rjain/ppzjj_100k/ppzjj_NLO_100k_ljets.npy')[:,0]
big_ljeteta = np.load('/oscar/data/mleblan6/rjain/ppzjj_100k/ppzjj_NLO_100k_ljets.npy')[:,1]
big_ljetphi = np.load('/oscar/data/mleblan6/rjain/ppzjj_100k/ppzjj_NLO_100k_ljets.npy')[:,2]

big_sljetpt = np.load('/oscar/data/mleblan6/rjain/ppzjj_100k/ppzjj_NLO_100k_sljets.npy')[:,0]
big_sljeteta = np.load('/oscar/data/mleblan6/rjain/ppzjj_100k/ppzjj_NLO_100k_sljets.npy')[:,1]
big_sljetphi = np.load('/oscar/data/mleblan6/rjain/ppzjj_100k/ppzjj_NLO_100k_sljets.npy')[:,2]

big_njets = np.load('/oscar/data/mleblan6/rjain/ppzjj_100k/ppzjj_NLO_100k_njets.npy')
```

```python
dijet_idx1 = np.where(big_njets > 1)[0]

big_ljet_sljet_dist = np.sqrt((big_ljeteta[dijet_idx1] - big_sljeteta[dijet_idx1])**2 + (big_ljetphi[dijet_idx1] - big_sljetphi[dijet_idx1])**2)
big_ljetpt_over_sljetpt = big_ljetpt[dijet_idx1] / big_sljetpt[dijet_idx1]

dijet_idx2 = np.where(big_njets > 0)[0]
big_z_ljet_dist = np.sqrt((big_ljeteta[dijet_idx2] - big_zeta[dijet_idx2])**2 + (big_ljetphi[dijet_idx2] - big_zphi[dijet_idx2])**2)
```

# PLOT WRT 100K SAMPLE

```python
#'ΔR(j_{1},j_{2})'
#'ΔR(Z,j_{1})'
#'Z p_{T}'
#'p_{T,j_{1}}/ p_{T,j_{2}}'
#'p_{T,ljet}/ p_{T,sljet}'

observable = big_ljet_sljet_dist #observable from 100k sample
logscale = True #whether y-axis should be log scaled or not
quantity = 'Z p_{T}' #string for labels and stuff is the observable
numbins = 20 #number of bins
logbins = False #whether bin widths should be log scaled or not
minbin = 25 #minimum bin width when log scaled
frac = 77 #fraction of reweighting used for caption in the plot

#100K ORIGINAL WEIGHT
Weight = big_weight

# LOAD 100K REWEIGHTED ARRAYS HERE
reweight1 = np.load(f'/oscar/data/mleblan6/jmarrinan/hp_reweights/100k_hp_emd_reweight_40gev_Rmax_beta0p5.npy')
reweight2 = np.load(f'/oscar/data/mleblan6/rjain/100k_reweight_bigR/100k_hp_emd_reweight_12p9gev_bigR.npy')
reweight3 = np.load(f'/oscar/data/mleblan6/jmarrinan/hp_reweights/100k_hp_emd_reweight_1p45gev_Rmax_beta2.npy')
reweight4 = np.load(f'/oscar/data/mleblan6/rjain/betainf/hardprocess_reweights.npy')[33]
reweight5 = np.load(f'/oscar/data/mleblan6/rjain/beta0/hardprocess_reweights.npy')[43]
```

```python
reweight_arrs = [reweight1, reweight2, reweight3, reweight4, reweight5]
zero_weight = 0

for k in range(len(reweight_arrs)):
    j = 0  # pointer for original array
    
    if len(reweight_arrs[k]) == 99996:
        new_array = []
        for i in range(100000):
            if i in mask:
                new_array.append(zero_weight)
            else:
                new_array.append(reweight_arrs[k][j])
                j += 1

        reweight_arrs[k] = new_array
    print(len(reweight_arrs[k]))

reweight1 = np.array(reweight_arrs[0])
reweight2 = np.array(reweight_arrs[1])
reweight3 = np.array(reweight_arrs[2])
reweight4 = np.array(reweight_arrs[3])
reweight5 = np.array(reweight_arrs[4])
```

```python
print('Reweighted fractions for hard process, showered, hadronization arrays respectively:')
print(1-len(np.where(reweight1 < 0)[0]) / len(np.where(Weight < 0)[0]))
print(1-len(np.where(reweight2 < 0)[0]) / len(np.where(Weight < 0)[0]))
print(1-len(np.where(reweight3 < 0)[0]) / len(np.where(Weight < 0)[0]))
print(1-len(np.where(reweight4 < 0)[0]) / len(np.where(Weight < 0)[0]))
print(1-len(np.where(reweight5 < 0)[0]) / len(np.where(Weight < 0)[0]))

print('----------------------------------------------------------------------')

print(f'Max of 100k observable is: {max(observable)}')
```

```python
if len(observable) == len(big_ljet_sljet_dist):
    Weight = big_weight[dijet_idx1]
    reweight1 = reweight1[dijet_idx1]
    reweight2 = reweight2[dijet_idx1]
    reweight3 = reweight3[dijet_idx1]
    reweight4 = reweight4[dijet_idx1]
    reweight5 = reweight5[dijet_idx1]

elif len(observable) == len(big_z_ljet_dist):
    Weight = big_weight[dijet_idx2]
    reweight1 = reweight1[dijet_idx2]
    reweight2 = reweight2[dijet_idx2]
    reweight3 = reweight3[dijet_idx2]
    reweight4 = reweight4[dijet_idx2]
    reweight5 = reweight5[dijet_idx2]


print('Length of observable and weight arrays. These must be the same:')
print(len(observable), ',', len(Weight))
print(len(observable), ',', len(reweight1))
print(len(observable), ',', len(reweight2))
print(len(observable), ',', len(reweight3))
print(len(observable), ',', len(reweight4))
print(len(observable), ',', len(reweight5))
```

```python
#create canvas
c1 = ROOT.TCanvas("c1", "Canvas", 800, 600)

#setup all the histograms
if logbins == False:    
    hist = ROOT.TH1D("Original", "My Histogram", numbins, min(observable) - min(observable) / 100, 
                     max(observable) + max(observable) / 100)
    
    hist2 = ROOT.TH1D("Reweight 1", "My Histogram", numbins, min(observable) - min(observable) / 100, 
                      max(observable) + max(observable) / 100)
    
    hist3 = ROOT.TH1D("Reweight 2", 'My Histogram', numbins, min(observable) - min(observable) / 100, 
                      max(observable) + max(observable) / 100)

    hist4 = ROOT.TH1D("Reweight 3", 'My Histogram', numbins, min(observable) - min(observable) / 100, 
                      max(observable) + max(observable) / 100)

    hist5 = ROOT.TH1D("Reweight 4", 'My Histogram', numbins, min(observable) - min(observable) / 100, 
                  max(observable) + max(observable) / 100)

    hist6 = ROOT.TH1D("Reweight 5", 'My Histogram', numbins, min(observable) - min(observable) / 100, 
                  max(observable) + max(observable) / 100)
    
elif logbins == True:
    log_bins = np.logspace(np.log10(min(observable) - min(observable) / 100), np.log10(max(observable) + max(observable) / 100), 
                           numbins + 1)

    pruned_bins = [log_bins[0]]
    for b in log_bins[1:]:
        if b - pruned_bins[-1] >= minbin:
            pruned_bins.append(b)
            
    log_bins_array = array('d', pruned_bins)    
    hist = ROOT.TH1D("Original", "My Histogram", len(log_bins_array)-1, log_bins_array)
    
    hist2 = ROOT.TH1D("Reweight 1", "My Histogram", len(log_bins_array)-1, log_bins_array)
    
    hist3 = ROOT.TH1D("Reweight 2", 'My Histogram', len(log_bins_array)-1, log_bins_array)

    hist4 = ROOT.TH1D("Reweight 3", 'My Histogram', len(log_bins_array)-1, log_bins_array)

    hist5 = ROOT.TH1D("Reweight 4", 'My Histogram', len(log_bins_array)-1, log_bins_array)

    hist6 = ROOT.TH1D("Reweight 5", 'My Histogram', len(log_bins_array)-1, log_bins_array)

for value, weight2 in zip(observable, Weight):
    hist.Fill(value, weight2)  # Fill with value and weight

for value, weight2 in zip(observable, reweight1):
    hist2.Fill(value, weight2)  # Fill with value and weight

for value, weight2 in zip(observable, reweight2):
    hist3.Fill(value, weight2)  # Fill with value and weight

for value, weight2 in zip(observable, reweight3):
    hist4.Fill(value, weight2)  # Fill with value and weight

for value, weight2 in zip(observable, reweight4):
    hist5.Fill(value, weight2)  # Fill with value and weight

for value, weight2 in zip(observable, reweight5):
    hist6.Fill(value, weight2)  # Fill with value and weight


hist.Scale(1.0 / hist.Integral())
hist2.Scale(1.0 / hist2.Integral())
hist3.Scale(1.0 / hist3.Integral())
hist4.Scale(1.0 / hist4.Integral())
hist5.Scale(1.0 / hist5.Integral())
hist6.Scale(1.0 / hist6.Integral())


ratio = hist.Clone()
ratio.Divide(hist)
ratio.SetLineColor(ROOT.kBlack)
ratio.SetMarkerColor(ROOT.kBlack)
ratio.SetStats(0)

ratio2 = hist2.Clone()
ratio2.Divide(hist)
ratio2.SetLineColor(ROOT.kRed)
ratio2.SetMarkerColor(ROOT.kRed)
ratio2.SetStats(0)
ratio2.SetLineWidth(2)

ratio3 = hist3.Clone()
ratio3.Divide(hist)
ratio3.SetLineColor(ROOT.kBlue)
ratio3.SetMarkerColor(ROOT.kBlue)
ratio3.SetStats(0)
ratio3.SetLineWidth(2)

ratio4 = hist4.Clone()
ratio4.Divide(hist)
ratio4.SetLineColor(ROOT.kOrange)
ratio4.SetMarkerColor(ROOT.kOrange)
ratio4.SetStats(0)
ratio4.SetLineWidth(2)

ratio5 = hist5.Clone()
ratio5.Divide(hist)
ratio5.SetLineColor(ROOT.kGreen)
ratio5.SetMarkerColor(ROOT.kGreen)
ratio5.SetStats(0)
ratio5.SetLineWidth(2)

ratio6 = hist6.Clone()
ratio6.Divide(hist)
ratio6.SetLineColor(ROOT.kMagenta)
ratio6.SetMarkerColor(ROOT.kMagenta)
ratio6.SetStats(0)
ratio6.SetLineWidth(2)

#top plot
pad1 = ROOT.TPad("pad1","pad1" ,0 ,0.3 ,1 ,1)

# for h in [hist0, hist, hist2, hist3, hist4]:
#     h.GetXaxis().SetRangeUser(min(observable), max(observable))



if logscale == True:
    pad1.SetLogy(True)
else:
    pad1.SetLogy(False)
pad1.Draw()
pad1.cd()
hist.Draw("P h")
hist2.Draw("P e1x0 same")
hist3.Draw('P e1x0 same')
hist4.Draw('P e1x0 same')
hist5.Draw('P e1x0 same')
hist6.Draw('P e1x0 same')

hist.SetLineColor(ROOT.kBlack)
hist.SetMarkerColor(ROOT.kBlack)
hist.SetLineWidth(2)

hist2.SetLineColor(ROOT.kRed)
hist2.SetMarkerColor(ROOT.kRed)
hist2.SetLineWidth(2)

hist3.SetLineColor(ROOT.kBlue)
hist3.SetMarkerColor(ROOT.kBlue)
hist3.SetLineWidth(2)

hist4.SetLineColor(ROOT.kOrange)
hist4.SetMarkerColor(ROOT.kOrange)
hist4.SetLineWidth(2)

hist5.SetLineColor(ROOT.kGreen)
hist5.SetMarkerColor(ROOT.kGreen)
hist5.SetLineWidth(2)

hist6.SetLineColor(ROOT.kMagenta)
hist6.SetMarkerColor(ROOT.kMagenta)
hist6.SetLineWidth(2)

hist.SetStats(0)

legend = ROOT.TLegend(0.6 ,0.68 ,0.8 ,0.88)
legend.AddEntry(hist ,"Original 100k Sample")
legend.AddEntry(hist6 , 'Hard Process #beta = 0')
legend.AddEntry(hist2 ,"Hard Process #beta = 0.5")
legend.AddEntry(hist3 , 'Hard Process #beta = 1')
legend.AddEntry(hist4 , 'Hard Process #beta = 2')
legend.AddEntry(hist5 , 'Hard Process #beta = #infty')

legend.SetLineWidth(0)
legend.SetTextSize(0.5)
legend.Draw("same")

hist.GetYaxis().SetTitle(f"d#sigma/d{quantity}")
hist.GetYaxis().SetTitleSize(0.08)
hist.GetYaxis().SetTitleOffset(0.4)

latex = ROOT.TLatex()
latex.SetNDC()          # coordinates normalized (0-1)
latex.SetTextSize(0.04) # font size
latex.SetTextColor(ROOT.kBlack)
latex.DrawLatex(0.63, 0.65, f'Reweighted fraction = ~{frac}%')  # (x, y, "text")

#bottom plot
c1.cd()
pad2 = ROOT.TPad("pad2","pad2" ,0 ,0.05 ,1 ,0.3)
pad2.Draw()
pad2.cd()
# for r in [ratio0, ratio, ratio2, ratio3, ratio4]:
#     r.GetXaxis().SetRangeUser(min(observable), max(observable))


pad1.SetBottomMargin(0)
pad2.SetTopMargin(0)
pad2.SetBottomMargin(0.32)


hist.SetTitle('')
hist.GetXaxis().SetTitle("")
hist.GetXaxis().SetTitleSize(0)

ratio.SetTitle("")
ratio.GetXaxis().SetLabelSize(0.12)
ratio.GetXaxis().SetTitleSize(0.18)
ratio.GetYaxis().SetLabelSize(0.1)
ratio.GetYaxis().SetTitleSize(0.07)
ratio.GetYaxis().SetTitle("Reweight / Original")
ratio.GetYaxis().SetTitleOffset(0.4)
ratio.GetYaxis().SetRangeUser(0.75 , 1.25)
ratio.GetYaxis().SetNdivisions(5, ROOT.kTRUE)
ratio.SetLineWidth(2)
ratio.GetXaxis().SetTitle(quantity)
ratio.GetXaxis().SetTitleOffset(0.7)


ratio.Draw('P')
ratio2.Draw('P e1x0 same')
ratio3.Draw('P e1x0 same')
ratio4.Draw('P e1x0 same')
ratio5.Draw('P e1x0 same')
ratio6.Draw('P e1x0 same')

c1.Draw()
```
